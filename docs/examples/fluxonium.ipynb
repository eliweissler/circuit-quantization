# Fluxonium

The fluxonium shunts a Josephson junction with a large inductance [Manucharyan et al., Science **326**, 113 (2009)](https://doi.org/10.1126/science.1175552).
The inductor closes a superconducting loop threaded by an external flux, and makes the node phase extended rather than compact.

```python
import sympy as sym
import circuit_quantization as cq
from circuit_quantization import Branch, Circuit, Loops, quantize
```

## Circuit

Energies are given in GHz and converted to SI values; the external flux is a reduced phase converted to webers.

```python
E_C, E_L, E_J = sym.symbols("E_C E_L E_J", positive=True)
varphi_ext = sym.Symbol("varphi_ext", real=True)

fluxonium = Circuit([
    Branch("J", 1, 0, C=cq.ghz_to_farad(E_C), EJ=E_J),
    Branch("L", 1, 0, L=cq.ghz_to_henry(E_L)),
], ground=0)
cq.draw_circuit_diagram(fluxonium, grounds="below")
```

## Loops

`from_branch_dict` names each loop by the branches around it; the first branch fixes its orientation.

```python
loops = Loops.from_branch_dict(fluxonium, {"f": ("J", "L")}, {"f": cq.phase_to_weber(varphi_ext)})
```

## Quantization

```python
result = quantize(fluxonium, loops, simplify=True)
assert result.coords.types == ("e",)
result
```

The single mode is extended (`e`): its phase enters both the cosine and the inductive energy, so it has no offset charge.
The external flux appears as a linear term; `result.ham` holds the same Hamiltonian as matrices.
